```python

# DAY 8 - NUMPY
# BASIC TO ADVANCED

import numpy as np

# 1. CREATE NUMPY ARRAYS

ages = np.array([25, 32, 45, 29, 51, 38, 42, 27, 60, 35])

weights = np.array([55, 68, 75, 62, 82, 70, 78, 58, 85, 65])

heights = np.array([160, 172, 175, 165, 180, 170, 178, 162, 182, 168])

blood_pressure = np.array([120, 130, 140, 125, 150, 135, 145, 118, 155, 128])

print("\n1. NUMPY ARRAYS")
print("Ages:", ages)
print("Weights:", weights)
print("Heights:", heights)
print("Blood Pressure:", blood_pressure)
```

```python
# 2. ARRAY INFORMATION

print("\n2. ARRAY INFORMATION")
print("Type:", type(ages))
print("Size:", ages.size)
print("Shape:", ages.shape)
print("Dimensions:", ages.ndim)
print("Data Type:", ages.dtype)

```

```python
# 3. INDEXING

print("\n3. INDEXING")
print("First Age:", ages[0])
print("Fourth Age:", ages[3])
print("Last Age:", ages[-1])
```

```python
# 4. SLICING


print("\n4. SLICING")
print("First Five Ages:", ages[:5])
print("Ages from Position 2 to 6:", ages[2:7])
print("First Three Ages:", ages[:3])
print("Last Three Ages:", ages[-3:])
```

```python
# 5. MATHEMATICAL OPERATIONS


print("\n5. MATHEMATICAL OPERATIONS")
print("Age + 5:", ages + 5)
print("Age - 5:", ages - 5)
print("Age * 2:", ages * 2)
print("Age / 2:", ages / 2)
print("Age Squared:", ages ** 2)
```

```python
# 6. OPERATIONS BETWEEN ARRAYS

array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([5, 10, 15, 20, 25])

print("\n6. ARRAY OPERATIONS")
print("Addition:", array1 + array2)
print("Subtraction:", array1 - array2)
print("Multiplication:", array1 * array2)
print("Division:", array1 / array2)
```

```python
# 7. STATISTICAL FUNCTIONS

print("\n7. STATISTICAL FUNCTIONS")
print("Sum:", np.sum(ages))
print("Mean:", np.mean(ages))
print("Median:", np.median(ages))
print("Minimum:", np.min(ages))
print("Maximum:", np.max(ages))
print("Standard Deviation:", np.std(ages))
print("Variance:", np.var(ages))
```

```python
# 8. FILTERING WITH CONDITIONS

print("\n8. FILTERING")
print("Ages Above 40:", ages[ages > 40])
print("Ages Below 30:", ages[ages < 30])
print("Ages Equal to 35:", ages[ages == 35])
print("Ages Between 30 and 50:", ages[(ages >= 30) & (ages <= 50)])
```

```python
# 9. np.where()

print("\n9. NP.WHERE")

age_category = np.where(ages >= 40, "Senior", "Adult")

print("Age Categories:", age_category)
```

```python
# 10. TWO-DIMENSIONAL ARRAY

patient_data = np.array([
    [101, 25, 55, 120],
    [102, 32, 68, 130],
    [103, 45, 75, 140],
    [104, 29, 62, 125],
    [105, 51, 82, 150]
])

print("\n10. TWO-DIMENSIONAL ARRAY")
print(patient_data)
print("Shape:", patient_data.shape)
print("Dimensions:", patient_data.ndim)
print("Total Elements:", patient_data.size)
```

```python
# 11. ACCESS ROWS AND COLUMNS

print("\n11. ROWS AND COLUMNS")
print("First Row:", patient_data[0])
print("Third Row:", patient_data[2])
print("Patient IDs:", patient_data[:, 0])
print("Patient Ages:", patient_data[:, 1])
print("Patient Weights:", patient_data[:, 2])
print("Blood Pressure:", patient_data[:, 3])
```

```python
# 12. RESHAPE


numbers = np.array([1, 2, 3, 4, 5, 6])

reshaped_array = numbers.reshape(2, 3)

print("\n12. RESHAPE")
print(reshaped_array)
```

```python
# 13. ARRAY CREATION FUNCTIONS

print("\n13. ARRAY CREATION")
print("Zeros:", np.zeros(5))
print("Ones:", np.ones(5))
print("Range:", np.arange(1, 11))
print("Even Numbers:", np.arange(2, 21, 2))
print("Five Equal Intervals:", np.linspace(0, 100, 5))
```

```python
# 14. RANDOM NUMBERS

np.random.seed(42)

random_numbers = np.random.randint(1, 100, 10)

print("\n14. RANDOM NUMBERS")
print(random_numbers)
```

```python
# 15. SORTING

numbers = np.array([50, 20, 80, 10, 40])

print("\n15. SORTING")
print("Original:", numbers)
print("Sorted:", np.sort(numbers))
print("Sorted Descending:", np.sort(numbers)[::-1])
```

```python
# 16. UNIQUE VALUES

drugs = np.array([
    "Paracetamol",
    "Metformin",
    "Paracetamol",
    "Ibuprofen",
    "Metformin",
    "Amoxicillin"
])

print("\n16. UNIQUE VALUES")
print(np.unique(drugs))
```

```python
# 17. CONCATENATE ARRAYS


a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

print("\n17. CONCATENATE")
print(np.concatenate((a, b)))
```

```python
# 18. AGGREGATION USING AXIS

sales = np.array([
    [1000, 2000, 3000],
    [1500, 2500, 3500],
    [1200, 2200, 3200]
])

print("\n18. AXIS OPERATIONS")
print("Sales Data:")
print(sales)
print("Column Totals:", np.sum(sales, axis=0))
print("Row Totals:", np.sum(sales, axis=1))
print("Column Averages:", np.mean(sales, axis=0))
print("Row Averages:", np.mean(sales, axis=1))
```

```python
# 19. MINIMUM AND MAXIMUM POSITION

print("\n19. MINIMUM AND MAXIMUM POSITIONS")
print("Minimum Position:", np.argmin(ages))
print("Maximum Position:", np.argmax(ages))
```

```python
# 20. CLIP VALUES


values = np.array([10, 25, 40, 65, 90])

print("\n20. CLIP VALUES")
print(np.clip(values, 20, 70))
```

```python
# 21. CHECK ANY AND ALL CONDITIONS

print("\n21. ANY AND ALL")
print("Any Age Above 60:", np.any(ages > 60))
print("All Ages Above 18:", np.all(ages > 18))
```

```python
# 22. SORTED PATIENT AGES

print("\n22. SORTED PATIENT AGES")
print(np.sort(ages))
```

```python
# 23. FINAL NUMPY SUMMARY

print("\n23. FINAL SUMMARY")
print("Total Patients:", ages.size)
print("Average Age:", np.mean(ages))
print("Average Weight:", np.mean(weights))
print("Average Height:", np.mean(heights))
print("Average Blood Pressure:", np.mean(blood_pressure))

print("\nNUMPY TASK COMPLETED")
```
